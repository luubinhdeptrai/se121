### STEP 6: Train

```
!python main.py \
  --mode train_fusion \
  --fusion_type gmu \
  --text_model_name uitnlp/visobert \
  --image_model_name swin_base_patch4_window7_224 \
  --epochs 20 \
  --batch_size 16 \
  --lr 1e-5 \
  --grad_accum_steps 2 \
  --patience 5 \
  --loss_fn auto_weight \
  --unfreeze_text_layers 1 \
  --unfreeze_image_layers 1 \
  --seed 42 \
  --use_amp \
  --exp_id EXP_060B_swinb_visobert_gmu_uncertainty \
  --exp_dir ./experiments
```
